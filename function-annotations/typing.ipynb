## 型エイリアス

type statementを利用して型を定義できる

```python
type Reward = list[int]
```

```python
def cal_reward(x: int) -> Reward:
    return [x**2]
```

```python
cal_reward(3)
```

```python
## 型エイリアスを明示的にすることもできる
from typing import TypeAlias

ABC: TypeAlias = list[str]
```

## NewType

- 新しい型を作成する  
- 元の型のサブクラス

```python
from typing import NewType
```

```python
NetworkID = NewType('NetworkID',int)
```

```python
NetworkID
```

```python
NetworkID(3141592)
```

```python
NetworkID("3141592")
```
